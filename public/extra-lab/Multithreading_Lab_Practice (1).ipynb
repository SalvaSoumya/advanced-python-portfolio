```python
import concurrent.futures
import time

start = time.perf_counter()

def do_something(seconds):
    print(f'sleeping {seconds} second (s)')
    time.sleep(seconds)
    return 'Done sleeping...'

with concurrent.futures.ThreadPoolExecutor() as executor:
    f1 = executor.submit(do_something, 1.5)
    f2 = executor.submit(do_something, 1.5)
    print(f1.result())
    print(f2.result())
    res=[executor.submit(do_something, 1.5) for _ in range(2)]
    # for _ in range(2):
    #     f1 = executor.submit(do_something, 1.5)

    for f in concurrent.futures.as_completed(res):
        print(f.result())
        
    # ffor _ in range(2):
finish = time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')
```

CONCURRENT FUTURES - 2

1. Creating many worker threads/ Thread pool using loop/list comprehension
2. understanding as_completed(results) - returns an iterator object which fetches the result as and when it is completed

```python
import concurrent.futures
import time

start = time.perf_counter()

def do_something(seconds):
    print(f'sleeping {seconds} second (s)')
    time.sleep(seconds)
    return 'Done sleeping...'

with concurrent.futures.ThreadPoolExecutor() as executor:
    results = [executor.submit(do_something, 1) for _ in range(10)]

    for f in concurrent.futures.as_completed(results):
        print(f.result())
    
finish = time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')
```

CONCURRENT FUTURES - 4

Using the built-in map() method

Difference: map does not hold a future object (only submit does that), it holds the results in the order they were started


```python
import concurrent.futures
import time

start = time.perf_counter()


def do_something(seconds):
    print(f'Sleeping {seconds} second(s)...')
    time.sleep(seconds)
    return f'Done Sleeping...{seconds}'

with concurrent.futures.ThreadPoolExecutor() as executor:
    secs = [5, 4, 3, 2, 1]
    results = executor.map(do_something, secs)

for result in results:
    print(result)

finish = time.perf_counter()

print(f'Finished in {round(finish-start, 2)} second(s)')
```

Future    → represents the task/result

results   → list containing multiple Futures

2. import concurrent.futures
3. After the function, create the ThreadPoolExecutor() in a context manager
4. As it is an IO bound task, Threads are effective - downloading high resolution images, observe that the time is took was far lesser compared to the earlier version
